# 02 - Visualizing Data with matplotlib 📊

A good chart can show in one second what a table of numbers hides.

**Pick the chart for the question:**
| Question | Chart |
|----------|-------|
| How does something change over time? | **line** |
| Are two numbers related? | **scatter** |
| How do categories compare? | **bar** |
| How are values spread out? | **histogram** |

```python
import pandas as pd
import matplotlib.pyplot as plt

penguins = pd.read_csv("data/penguins.csv").dropna()
aapl = pd.read_csv("data/AAPL.csv", parse_dates=["Date"], index_col="Date")
```

## 1. Line plot: change over time
The basic recipe is always the same: **plot → label → show**.

```python
plt.figure(figsize=(10, 4))
plt.plot(aapl.index, aapl["Close"])
plt.title("Apple (AAPL) closing price")
plt.xlabel("Date")
plt.ylabel("Price (USD)")
plt.grid(alpha=0.3)
plt.show()
```

## 2. Scatter plot: are two variables related?
Do penguins with longer flippers weigh more?

```python
plt.figure(figsize=(7, 5))
plt.scatter(penguins["flipper_length_mm"], penguins["body_mass_g"], alpha=0.6)
plt.xlabel("Flipper length (mm)")
plt.ylabel("Body mass (g)")
plt.title("Longer flippers, heavier penguins")
plt.show()
```

### Color by category: loop over the groups and add a **legend**

```python
plt.figure(figsize=(7, 5))
for species in penguins["species"].unique():
    group = penguins[penguins["species"] == species]
    plt.scatter(group["bill_length_mm"], group["bill_depth_mm"], label=species, alpha=0.7)
plt.xlabel("Bill length (mm)")
plt.ylabel("Bill depth (mm)")
plt.title("Each species has a different bill shape")
plt.legend()
plt.show()
```

☝️ Notice how the three species form **clusters**? Next week, a machine-learning model will learn those clusters to *predict* a penguin's species from its measurements.

## 3. Bar chart: comparing categories

```python
counts = penguins["species"].value_counts()
plt.bar(counts.index, counts.values, color=["tab:blue", "tab:orange", "tab:green"])
plt.ylabel("Number of penguins")
plt.title("Penguins per species")
plt.show()
```

```python
avg_mass = penguins.groupby("species")["body_mass_g"].mean().sort_values()
plt.barh(avg_mass.index, avg_mass.values)     # barh = horizontal bars
plt.xlabel("Average body mass (g)")
plt.show()
```

## 4. Histogram: how are values distributed?

```python
plt.hist(penguins["body_mass_g"], bins=20, edgecolor="black")
plt.xlabel("Body mass (g)")
plt.ylabel("Count")
plt.title("Distribution of penguin body mass")
plt.show()
```

```python
# Overlapping histograms (alpha = transparency)
for species in ["Adelie", "Gentoo"]:
    plt.hist(penguins[penguins["species"] == species]["flipper_length_mm"], bins=15, alpha=0.6, label=species)
plt.xlabel("Flipper length (mm)")
plt.legend()
plt.show()
```

## 5. Several charts in one figure: subplots

```python
fig, axes = plt.subplots(1, 3, figsize=(14, 4))

axes[0].hist(penguins["bill_length_mm"], bins=20)
axes[0].set_title("Bill length")

axes[1].scatter(penguins["bill_length_mm"], penguins["flipper_length_mm"], s=10)
axes[1].set_title("Bill vs flipper")

penguins["island"].value_counts().plot(kind="bar", ax=axes[2])   # pandas can plot too!
axes[2].set_title("Penguins per island")

plt.tight_layout()
plt.show()
```

## 6. The pandas shortcut: `df.plot(...)`
pandas wraps matplotlib, so many charts take one line.

```python
monthly = aapl["Volume"].groupby(aapl.index.to_period("M")).sum()
monthly.plot(kind="bar", figsize=(12, 3), title="AAPL monthly trading volume")
plt.xticks([])      # too many labels to read -- hide them
plt.show()
```

## 7. Saving a chart

```python
import os
os.makedirs("output", exist_ok=True)
penguins.plot(kind="scatter", x="flipper_length_mm", y="body_mass_g", title="Saved chart")
plt.savefig("output/penguins.png", dpi=150, bbox_inches="tight")
plt.show()
```
