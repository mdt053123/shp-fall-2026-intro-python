## NumPy recap
Let's review today's NumPy with a tiny experiment: **roll two dice 100,000 times**.

```python
import numpy as np

rng = np.random.default_rng(seed=0)
die1 = rng.integers(1, 7, size=100_000)
die2 = rng.integers(1, 7, size=100_000)
totals = die1 + die2          # vectorized: no loop!
totals[:10]
```

```python
# How often does each total appear?
values, counts = np.unique(totals, return_counts=True)
for v, c in zip(values, counts):
    print(f"{v:>2}: {'#' * (c // 500)} {c / len(totals):.1%}")
```

## A first taste of plotting
Next week we'll learn **matplotlib** properly. For now, see how a chart shows up right in the notebook:

```python
import matplotlib.pyplot as plt

plt.bar(values, counts)
plt.xlabel("Total of two dice")
plt.ylabel("How many times")
plt.title("100,000 rolls of two dice")
plt.show()
```
